```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
insar = pd.read_csv(
    "../../data/raw/insar/insar.csv"
)

seismic = pd.read_csv(
    "../../data/raw/seismic/seismic.csv"
)

print(
    "InSAR:",
    insar.shape
)

print(
    "Seismic:",
    seismic.shape
)
```

```python
print("InSAR columns:")
print(insar.columns.tolist())

print("\nSeismic columns:")
print(seismic.columns.tolist())
```

```python
print(
    insar.info()
)

print(
    seismic.info()
)
```

```python
print(
    insar.isnull().sum()
)

print(
    seismic.isnull().sum()
)
```

```python
print(
    insar.describe()
)
```

```python
print(
    seismic.describe()
)
```

```python
numeric_columns = insar.select_dtypes(
    include="number"
).columns

for column in numeric_columns:

    plt.figure(figsize=(10, 4))

    plt.plot(
        insar[column]
    )

    plt.title(
        f"InSAR - {column}"
    )

    plt.xlabel(
        "Observation"
    )

    plt.ylabel(
        column
    )

    plt.show()
```

```python
numeric_columns = seismic.select_dtypes(
    include="number"
).columns

for column in numeric_columns:

    plt.figure(figsize=(10, 4))

    plt.plot(
        seismic[column]
    )

    plt.title(
        f"Seismic - {column}"
    )

    plt.xlabel(
        "Observation"
    )

    plt.ylabel(
        column
    )

    plt.show()
```
